```python
import json
import shutil
import os
from tqdm import tqdm

JSON_FILE = '/home/felipep/Descargas/mpii_human_pose_v1/annotations_trainval2017/annotations/instances_train2017.json'
IMAGES_DIR = '/home/felipep/Descargas/mpii_human_pose_v1/annotations_trainval2017/train2017'
OUTPUT_DIR = 'dataset_final/negatives'
TARGET_COUNT = 4500

os.makedirs(OUTPUT_DIR, exist_ok=True)

print(f"Cargando {JSON_FILE} (esto puede tardar porque es grande)...")
with open(JSON_FILE, 'r') as f:
    data = json.load(f)

# encontrar el ID de la categoría 'person'
person_id = None
for cat in data['categories']:
    if cat['name'] == 'person':
        person_id = cat['id']
        break

if person_id is None:
    print("Error: No se encontró la categoría 'person'")
    exit()

print(f"ID de Categoría 'person': {person_id}")

# identificar imágenes que si tienen personas
print("Identificando imágenes con personas...")
images_with_people = set()
for ann in data['annotations']:
    if ann['category_id'] == person_id:
        images_with_people.add(ann['image_id'])

print(f"Total imágenes con personas ignoradas: {len(images_with_people)}")

# filtrar y copiar imágenes que NO tienen personas
count = 0
print(f"Extrayendo negativos a '{OUTPUT_DIR}'...")

# iterar sobre la lista de archivos de imagen disponibles
for img_info in tqdm(data['images']):
    img_id = img_info['id']
    file_name = img_info['file_name']
    
    # si el ID no está en el set de personas, es un negativo válido
    if img_id not in images_with_people:
        src_path = os.path.join(IMAGES_DIR, file_name)
        dst_path = os.path.join(OUTPUT_DIR, file_name)
        
        if os.path.exists(src_path):
            shutil.copy(src_path, dst_path)
            count += 1
        
        if count >= TARGET_COUNT:
            break

print("="*30)
print(f"PROCESO COMPLETADO")
print(f"Imágenes negativas guardadas: {count}")
print(f"Ubicación: {os.path.abspath(OUTPUT_DIR)}")
```
